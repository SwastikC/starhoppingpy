```python
import os
import glob
import subprocess
from starhoppingpy import sh_sort as shs
```

```python
if __name__ == '__main__':
    data_dir = "/path/to/raw"
    dest_dir = "/path/to/processed"
    msgfile = "msglog.txt"
    
    results = shs.get_file_list(data_dir)
    file_entries = [shs.parse_dfits_line(row) for row in results]
    datasets = shs.group_datasets(file_entries)
    print("Grouped datasets:")
    for ds_id, entries in datasets.items():
        print(f"Dataset ID: {ds_id} has {len(entries)} files")
        dataset_dir = shs.process_science_files(ds_id, entries, data_dir, dest_dir, msgfile)
        science_entry = entries[0]
        shs.process_calibration_flat(file_entries, science_entry, data_dir, dataset_dir, msgfile)
        shs.process_calibration_sky(file_entries, science_entry, data_dir, dataset_dir, msgfile)
        shs.process_calibration_dark(file_entries, science_entry, data_dir, dataset_dir, msgfile)
        shs.process_calibration_center(file_entries, science_entry, data_dir, dataset_dir, msgfile)
        shs.process_calibration_flux(file_entries, science_entry, data_dir, dataset_dir, msgfile)
        if science_entry["arm"].upper() == "IFS":
            shs.process_calibration_ifscal(file_entries, science_entry, data_dir, dataset_dir, msgfile)
            shs.process_calibration_ifscal2(file_entries, science_entry, data_dir, dataset_dir, msgfile)
    
    # Print a summary of science file counts
    shs.print_science_summary(dest_dir)
```
